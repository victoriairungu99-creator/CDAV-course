```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Load datsets from the seaborn library(seaborn comes with in built datasets)tips =  sns.load_dataset('tips')
tips=sns.load_dataset("tips")
penguins = sns.load_dataset("penguins")
flights= sns.load_dataset("flights")
fmri = sns.load_dataset("fmri")
```

```python
#Using matplotlib to set the size of the axis
fig,  axes = plt.subplots(2,4, figsize=(16,4))
ax = axes.flat #to help you get the positions of the diff charts
```

```python
fig,  axes = plt.subplots(2,4, figsize=(16,8))
ax = axes.flat

sns.scatterplot(data=tips, x= "total_bill", y="tip",
                hue="smoker", palette="pastel", ax=ax[0])
ax[0].set_title("scatter plot")

#lineplot-aggregated trend over discrete x
sns.lineplot(data= tips, x="size", y="tip",
             hue="smoker", marker="o", palette="pastel", ax=ax[1])
ax[1].set_title("Lineplot(mean tips )")

sns.barplot(data=tips, x="day", y="tip",
            hue="smoker", palette="pastel",ax=ax[2])
ax[2].set_title("barplot")

sns.boxplot(data=tips, x="day", y="tip",
            hue="smoker", palette="pastel",ax=ax[3])
ax[3].set_title("boxplot")

sns.violinplot(data=tips, x="day", y="tip",
            hue="smoker",split=True,inner="quartile", palette="pastel",ax=ax[4])
ax[4].set_title("violinplot")

sns.histplot(data=tips, x="tip", hue="smoker",bins=20,
              multiple="stack", palette="pastel", ax=ax[5])
ax[5].set_title("histplot")

sns.kdeplot(data=tips,x="total_bill",hue="smoker",
            fill=True,common_norm=False,alpha=0.5,ax=ax[6])
ax[6].set_title("kdeplot")

#Remove the extra empty subplot
fig.delaxes(ax=ax[7])

plt.tight_layout()
plt.show()


```

```python
#relplot(relational plot):split panels into rows and columns
#shows relationship bteween numerical variables
#Column is for choosing the column that you want that has diff variations
sns.relplot(data=tips,x="total_bill",y="tip",col="time",hue="smoker")
plt.show()
```

```python
sns.displot(data=penguins,x="flipper_length_mm",col="species", kde=True)
plt.show()
```

```python
#catplot- for comparison btwn categories
sns.catplot(data=tips,x="day",y="total_bill",col="time", kind="box")
plt.show()
```

```python
#Regression plot(loine of best fit)-draws a line whre a lot of data points are
#draws the line using the concept of y=mx+b where m is the slope,x is the value of x and b is the point where the reg line cuts at the y axis
sns.regplot(data=tips,x="total_bill",y="tip");plt.show()
```
