For boolean masking in numpy and is (&), or is(|), not is(~)

```python
import numpy as np
scores=np.array([98,70,56,97,46])
scores[~(scores>95)&(scores>60)|(scores<50)]
```

```python
def adder(num):
  return num+5

```

```python
uadder=np.vectorize(adder) #np.vectorize converts our defined function "adder" into a universal function(ufunc) to "uadder"
```

```python
scores=np.array([90,45,67])#without looping, it adds 5 to all numbers in array
uadder(scores)
```

```python
uadder2=np.frompyfunc(adder,1,1)#with this method of np.frompyfunc, the values returned are always objects
uadder2(scores)
```

```python
def stat(num1,num2):
  return(num1+num2, (num1+num2)/2)
```

```python
ustat=np.frompyfunc(stat,2,2)
arr_1=np.array([1,2,3])
arr_2=np.array([2,3,4])
ustat(arr_1,arr_2)
```
