```python
#Create a function that can take either a list or array and then gets the size, shape, data type and no.of dimensions
import numpy as np
def array_info(data):
  data=np.array(data)
  print("Number of dimensions: ",data.ndim)
  print("Shape: ",data.shape)
  print("Size: ",data.size)
  print("Data type: ",data.dtype)

my_list=([[18,16],[20,22]])
array_info(my_list)


```
