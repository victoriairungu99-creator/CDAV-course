```python
uadder=np.vectorize(adder) #np.vectorize converts our defined function "adder" into a universal function(ufunc) to "uadder"
```

```python
scores=np.array([90,45,67])#without looping, it adds 5 to all numbers in array
uadder(scores)
```

```python
uadder2=np.frompyfunc(adder,1,1)#with this method of np.frompyfunc, the values returned are always objects
uadder2(scores)
```

Create a function(summary)that displays the mean,median,std,var,dtype,size,shape of an array and returns the corresponding grade

```python
import numpy as np

def grade(score):
    if score >= 70:
        return "A"
    elif score >= 60:
        return "B"
    elif score >= 50:
        return "C"
    elif score >= 40:
        return "D"
    else:
        return "E"

# Vectorize the grading function
grade_vector = np.vectorize(grade)

def summary(data):
    data = np.array(data)

    mean = np.mean(data)

    print("Mean:", mean)
    print("Median:", np.median(data))
    print("Standard Deviation:", np.std(data))
    print("Variance:", np.var(data))
    print("Data Type:", data.dtype)
    print("Size:", data.size)
    print("Shape:", data.shape)

    # Put the mean into an array so vectorize can process it
    return grade_vector(np.array([mean]))[0]

marks = [80, 75, 90, 60, 70]

print("Grade:", summary(marks))
```

#A retail store tracks daily sales revenue across a 30-day month. You need to analyze sales trends, find peak performance days, filter out slow days, and flag anomalous sales spikes (outliers).

array([4374, 2060, 2494, 2330, 2295, 4292, 2838, 3369, 1666, 2438, 1530,
       2682, 8500, 4371, 4119, 1330, 2885, 1969, 3591, 2715, 4053, 3633,
       2415, 2155, 3524, 9200, 1659, 1221, 3500, 1947])

```python
import numpy as np
sales_rev=np.array([4374, 2060, 2494, 2330, 2295, 4292, 2838, 3369, 1666, 2438, 1530,      #create a numpy array of the sales revenue array to enable to get statistical data
       2682, 8500, 4371, 4119, 1330, 2885, 1969, 3591, 2715, 4053, 3633,
       2415, 2155, 3524, 9200, 1659, 1221, 3500, 1947])
print("Mean sales: ", np.mean(sales_rev))
print("Median sales: ", np.median(sales_rev))
print("Highest sales revenue : ", np.max(sales_rev))
print("Peak performance days: ", np.argmax(sales_rev)+1) #we add one because argmax function starts counting positions from 0 and days in a month begin with 1
print("Slow days: ",sales_rev[sales_rev<2500]) #filtering out sales that are below 2500
sales_rev[(sales_rev>=2500)] #keeping sales that are above or equal to 2500
q1=np.percentile(sales_rev,25) #getting the 25th percentile
q3=np.percentile(sales_rev,75) #getting the 75th percentile
iqr=q3-q1#to see how they are spread in the middle
lower=q1-1.5*iqr#for giving some allowance  for lower values
upper=q3+1.5*iqr#for giving some allowance for higher values

outliers=sales_rev[(sales_rev<lower)|(sales_rev>upper)] #the values that are below the lower range and values that are above the upper range
print("Outliers: ",outliers)

print("Peak sales revenue : ", sales_rev[sales_rev>q3])
print("Peak sales days : ", np.where(sales_rev>q3)[0]+1) #we put the zero because np.where always gives a tuple therefore the zero instructs python to print the first index of the tuple whiuch is the array list then ads 1 to each value



```
