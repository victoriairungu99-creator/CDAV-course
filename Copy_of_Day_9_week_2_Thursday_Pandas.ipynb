```python
import matplotlib.pyplot as plt #for charts
sales=df.groupby("county")["amount"].sum().sort_values()
sales.plot(kind="barh") #horizontal  bars
plt.show()
```

```python
import pandas as pd
df = pd.read_csv("/content/kenya_transactions_messy.csv")

df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
df = df.rename(columns={'amount_paid': 'amount'})

clean = (df['amount'].astype(str)
                     .str.replace(',', '', regex=False)
                     .str.replace(' KES', '', regex=False)
                     .str.strip())
df['amount'] = pd.to_numeric(clean, errors='coerce')

df["amount"].plot(kind="box")
df.boxplot(column="amount",by="county")
```
