```python
import pandas as pd
df=pd.read_csv("/content/kenya_transactions_messy.csv")
print("Before: ",df.columns.tolist()) #shows you how the columns look to see if there is any error like space before or after punctuation marks
```

```python
#makes the necessary changes
df.columns=(
    df.columns.str.strip()
    .str.lower()
    .str.replace(' ','_')
)
```

```python
#Verify
print("After: ",df.columns.tolist())
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
df = df.rename(columns={'amount_paid': 'amount'})
df.columns          # clean names, after
```

```python
clean = (df['amount'].astype(str)
                     .str.replace(',', '', regex=False)
                     .str.replace(' KES', '', regex=False)
                     .str.strip())
df['amount'] = pd.to_numeric(clean, errors='coerce')
df['amount'].dtype        # now a real number
```

```python
#errors='coerce' - Turns anything that is not numeric to NaN or a misiing value
```

```python
df["county"]=df["county"].str.strip().str.title()
```

```python
df.groupby("county")["amount"].sum().sort_values(ascending= True)
```

```python
df.groupby("county")["amount"].max()
```

```python
df.groupby("county")["amount"].agg(["count","sum","mean"])#.agg displays a lot of calculations at once
```
