```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
rng=np.random.default_rng(42)
n=12
df=pd.DataFrame({
    "month": pd.date_range("2024-01-01",
                           periods=n,
                           freq="MS").strftime("%b"),
    "sales": rng.integers(50,100,n),
    "cost":  rng.integers(20,60,n),
    "region": rng.choice(["East", "West"], n),
})
df["profit"]=df["sales"]-df["cost"] #creating a new column called profit
df.head()
```

```python
#Plot one column against another
fig,ax=plt.subplots()
ax.plot(df["month"],df["sales"])
plt.show()

```

```python
#comparing one solo variable with other several variables
fig,ax=plt.subplots()
ax.plot(df["month"],df["sales"])
ax.plot(df["month"],df["cost"])
ax.plot(df["month"],df["profit"])
plt.show()
```

```python
#to differentiate the various variables with colors
fig,ax=plt.subplots()
ax.plot(df["month"],df["sales"], color="crimson")
ax.plot(df["month"],df["cost"], color="steelblue")
ax.plot(df["month"],df["profit"], color="seagreen")
plt.show()
```

```python
#to show the various variables in different linestyles so as to distinguish them
#the ax.legend() method is used as a key for the axes/plot
#the legend gets its information from the label argument so you have to include it
fig,ax=plt.subplots()
ax.plot(df["month"],df["sales"],linestyle="-",
        linewidth=2,marker="o",label="sales")
ax.plot(df["month"],df["cost"],linestyle="--",
        linewidth=2,marker="s",label="cost")
ax.plot(df["month"],df["profit"],linestyle=":",
        linewidth=2,marker="^",label="profit")
ax.set_xlabel("Months")
ax.set_ylabel("Amount")
ax.set_title("Performance of year 2025")
ax.legend()
plt.show()
```

```python
#Bar chart of sales and another one for profit but in one figure
x=df["month"]
y1=df["sales"]
y2=df["profit"]
fig,axes=plt.subplots(1,2,figsize=(8,5))
axes[0].barh(x,y1);axes[0].set_title("Sales of 2025", fontstyle="italic", fontsize=13);axes[0].set_xlabel("Amount");axes[0].set_ylabel("Months")
axes[1].barh(x,y2);axes[1].set_title("Profit of 2025", fontstyle="italic", fontsize=13);axes[1].set_xlabel("Amount");axes[1].set_ylabel("Months")
fig.suptitle("Performance of year 2025",y=1,fontsize=16,fontweight="bold") #use super title(fig.suptitle for the title of the whole canvas)
plt.tight_layout();plt.show()
```

```python
#Mixture of two visuals in one figure
x=df["month"]
y1=df["sales"]
y2=df["profit"]
fig,axes=plt.subplots(1,2,figsize=(8,5))
axes[0].bar(x,y1);axes[0].set_title("Sales of 2025", fontstyle="italic", fontsize=13);axes[0].set_ylabel("Amount");axes[0].set_xlabel("Months")
axes[1].plot(x,y2);axes[1].set_title("Profit of 2025", fontstyle="italic", fontsize=13);axes[1].set_ylabel("Amount");axes[1].set_xlabel("Months")
fig.suptitle("Performance of year 2025",y=1,fontsize=16,fontweight="bold")
plt.tight_layout();plt.show()
```

```python
x=np.arange(len(df))
fig,ax=plt.subplots()
ax.bar(x-0.2,df["sales"],0.4,
       label="sales",color="darkblue")
ax.bar(x+0.2,df["cost"],0.4,
       label="cost",color="rosybrown")
ax.set_xticks(x,df["month"])
ax.set_xlabel("Months")
ax.set_ylabel("Sales and cost")
ax.set_title("Sales and Costs of year 2025", fontsize=16,fontweight="bold")
ax.legend()
plt.show()
```

```python
#Stacked bars
#The whole stacked bar is the sum of the two variables in it
fig,ax=plt.subplots()
ax.bar(df["month"],df["sales"],label="sales")
ax.bar(df["month"],df["cost"],bottom=df["sales"],label="cost")
ax.set_title("Stacked bars");ax.legend()
plt.show()
```

```python
df["quarter"]="Q"+(np.arange(len(df))//3+1).astype(str)
by_q=df.groupby("quarter")["sales"].sum()
colors=["#4C72B0","#DD8452","#55A868","#C44E52"]

fig,ax=plt.subplots()
ax.pie(by_q,labels=by_q.index,
       colors=colors,                 #custom slice colors
       autopct="%1.0f%%",             #show percentages
       startangle=90,                 #rotate the first slice to the top
       explode=[0.10,0,0,0],          #pull out the first slice to make it distinct
       shadow=True)

ax.set_title("Sales share per quarter")
plt.show()

```

```python
fig,ax=plt.subplots()
ax.pie(by_q,labels=by_q.index,
       colors=colors,
       autopct="%1.0f%%",
       startangle=90,
       wedgeprops=dict(width=0.4,edgecolor="white"))#width=donut

ax.text(0,0,"Sales",ha="center",va="center",
        fontsize=14,fontweight="bold")
ax.set_title("Donut with a centre label")
plt.show()
```

```python
rng2=np.random.default_rng(7)
products=["Laptop","Phone","Tablet","Watch"]
quarters=["Q1,","Q2","Q3","Q4"]
sales=pd.DataFrame(rng2.integers(20,100,size=(4,4)),
                   index=products,columns=quarters)
```

```python
#Customised heatmap
fig,ax=plt.subplots()
im=ax.imshow(sales,cmap="YlOrRd")
fig.colorbar(im,label="units sold")
ax.set_xticks(range(len(sales.columns)),sales.columns)
ax.set_xticks(range(len(sales.index)),sales.index)
for i in range(sales.shape[0]):
  for j in range(sales.shape[1]):
    ax.text(j,i,sales.iloc[i,j],ha="center",va="center")
ax.set_title("Quarterly sales by product")
plt.show()


```

```python
#impoprt library
from wordcloud import WordCloud
```

```python
text=("data visualization in python eg matplotlib seaborn pyplot\n"
      "a design story shows insight trend and pattern\n"
      "charts are a good way of showing insights and anomalies\n"
      "a bar chart is a chart that shows comparison\n"
      "a line chart is a chart that shows trend\n"
      "dashboards are useful for storyteling\n"
)

wc=WordCloud().generate(text)
fig,ax=plt.subplots()
ax.imshow(wc,interpolation="bilinear") #bilinear is for giving a clear view of the words
ax.axis("off") #to hide the x axis coz we do not need it in this library
plt.show()
```

```python
from wordcloud import STOPWORDS
print(len(STOPWORDS)) #how many
print(sorted(STOPWORDS)) #arrange them in alphabetical order
```

```python
#Specifying what stop words and any other words that you want to add to be excluded
wc=WordCloud(width=700,height=350,
             background_color="white",
             colormap="viridis",
             stopwords={"shows",'and',"a","is"}).generate(text)
fig,ax=plt.subplots(figsize=(7,3.5))
ax.imshow(wc,interpolation="bilinear");ax.axis("off")
plt.show()


```
